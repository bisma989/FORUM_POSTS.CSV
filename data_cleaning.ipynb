```python
df = pd.read_csv('../forum_posts.csv')
df.head()
```

```python
def clean_my_text(text):
    if isinstance(text, str):
        # 1. HTML Tags hatao (jaise <p>, <div>)
        text = re.sub(r'<[^>]+>', '', text)
        
        # 2. Internet ke links hatao (jaise http, www)
        text = re.sub(r'https?://\S+|www\.\S+', '', text)
        
        # 3. Emojis hatao
        text = text.encode('ascii', 'ignore').decode('ascii')
        
        # 4. Faltu khali jagah saaf karo
        text = " ".join(text.split())
        
        return text
    return text
```

```python
# 1. Computer khud hi file ke column ka naam check karega (Chahe Text ho ya Quote)
column_naam = df.columns[0] if df.columns[0] != 'id' else df.columns[1]

# 2. Safai chalana
df['clean_text'] = df[column_naam].apply(clean_my_text)

# 3. Dono columns ko screen par dekhna
df[[column_naam, 'clean_text']].head()
```
